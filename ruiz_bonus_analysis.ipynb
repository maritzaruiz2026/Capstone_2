```python
sales_detail = pd.merge(
    store_sales,
    store_detail,
    on='Store ID',
    how="left")

revenue_by_region = (
    sales_detail 
        .groupby("Region")['Sale Amount']
        .sum()
        .reset_index(name="Total Revenue")
        .sort_values("Total Revenue", ascending=False))

revenue_by_region
```
